```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError(f"score 超出范围: {score}")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for sc in (95, 90, 85, 72, 60, 59):
    print(f"score_level({sc}) = {score_level(sc)}")
try:
    score_level(101)
except ValueError as e:
    print("score_level(101) 抛异常 ->", type(e).__name__, ":", e)

total_for = 0
for k in range(1, 101):
    total_for += k

total_while = 0
k = 1
while k <= 100:
    total_while += k
    k += 1
print("for 求和   =", total_for)
print("while 求和 =", total_while)
print("两种实现一致:", total_for == total_while)
```

```python
def factorial_rec(n):
    return 1 if n <= 1 else n * factorial_rec(n - 1)


def factorial_iter(n):
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


for n in (0, 1, 5, 10):
    print(f"{n:>2}! 递归 = {factorial_rec(n):<8} 迭代 = {factorial_iter(n):<8} "
          f"一致: {factorial_rec(n) == factorial_iter(n)}")


def summary(values):
    return min(values), max(values), sum(values) / len(values)


lo, hi, avg = summary([85, 58, 92, 76, 60, 89])
print(f"元组解包: 最小 = {lo}, 最大 = {hi}, 平均 = {avg:.2f}")


def sum_all(*args):
    return sum(args)


print("sum_all()           =", sum_all())
print("sum_all(5)          =", sum_all(5))
print("sum_all(1, 2, 3, 4) =", sum_all(1, 2, 3, 4))
```

```python
import os
import pickle

cwd = os.getcwd()
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("工作目录:", cwd)
print("数据目录:", data_dir)
print("目录已存在:", os.path.isdir(data_dir))

content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
txt_path = os.path.join(data_dir, "students.txt")
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(content)
print()
print("已写入:", txt_path)
```

```python
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量应为2，实际 {len(parts)}")
    name, score_str = parts[0].strip(), parts[1].strip()
    if name == "":
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是整数: {score_str!r}")
    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0-100: {score}")
    return {"name": name, "score": score}


records, errors = [], []
with open(txt_path, encoding="utf-8") as f:
    f.readline()                      # 跳过表头
    for lineno, line in enumerate(f, start=2):
        if not line.strip():
            continue
        try:
            records.append(parse_student(line))
        except ValueError as e:
            errors.append((lineno, str(e)))

print("有效记录:")
for r in records:
    print("  ", r)
print("无效记录:")
for lineno, msg in errors:
    print(f"   第 {lineno} 行: {msg}")

valid_avg = sum(r["score"] for r in records) / len(records)
valid_rate = sum(1 for r in records if r["score"] >= 60) / len(records)
print()
print(f"有效 {len(records)} 条，无效 {len(errors)} 条")
print(f"平均分 {valid_avg:.2f}，及格率 {valid_rate:.1%}")
```

```python
report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as f:      # w：清空后重写
    f.write(f"有效记录数: {len(records)}\n")
    f.write(f"无效记录数: {len(errors)}\n")
    f.write(f"平均分: {valid_avg:.2f}\n")
    f.write(f"及格率: {valid_rate:.1%}\n")
print("--- 'w' 模式写入后 ---")
with open(report_path, encoding="utf-8") as f:
    print(f.read(), end="")

with open(report_path, "a", encoding="utf-8") as f:      # a：追加，不覆盖
    f.write("追加的一行\n")
print("--- 'a' 模式追加后 ---")
with open(report_path, encoding="utf-8") as f:
    print(f.read(), end="")
```

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student({self.name}, {self.score}, 及格={self.is_passed()})"


class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent({self.name}, {self.score}, "
                f"主题={self.research_topic}, 及格={self.is_passed()})")


# 为实验4的有效记录逐个建对象
for r in records:
    print(Student(r["name"], r["score"]))

print()
g = GraduateStudent("Alice", 85, "机器学习")
print(g)
print("g.is_passed() =", g.is_passed())
print("isinstance(g, Student) =", isinstance(g, Student))
```

```python
import copy


def make_original():
    return {"name": "Alice", "scores": [85, 90]}


direct = make_original()
alias = direct                 # 直接赋值：同一对象
shallow = copy.copy(direct)    # 浅拷贝：外层新 dict，内层 list 共享
deep = copy.deepcopy(direct)   # 深拷贝：完全独立

direct["scores"].append(100)
print("修改原始对象后:")
print("  直接赋值 alias  ->", alias["scores"])
print("  浅拷贝 shallow  ->", shallow["scores"])
print("  深拷贝 deep     ->", deep["scores"])
print("  深拷贝仍等于原始内容:", deep == make_original())
```
