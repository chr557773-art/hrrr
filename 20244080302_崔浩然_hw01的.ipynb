2. 定义⼀个名为 goat_pad 的函数，它接受⼀个字符串作为其唯⼀参数，并打印该字

欧⼏⾥得算法
欧⼏⾥得算法是⼀种寻找两个数的最⼤公约数（GCD）的⽅法。回想⼀下，两个数m和n
的最⼤公约数是同时能整除m和n的最⼤数。
1. 实现⼀个名为gcd的函数，它接受两个整数作为其参数并返回它们的GCD。你可以
假设两个输⼊都是整数，所以你不需要在你的函数中包含任何错误检查。
2. 使⽤你的函数评估以下数字对的GCD：
   1.(2018, 2019)
   2.(1200, 300)
   3.(5040, 60)
3. 如果⼀个或两个参数是负数，你的函数会做什么？这种⾏为有意义吗？



```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

for a, b in [(2018, 2019), (1200, 300), (5040, 60)]:
    print(f"gcd({a}, {b}) = {gcd(a, b)}")

print("负参数测试：")
for a, b in [(2018, -2019), (-2018, 2019), (-1200, -300)]:
    print(f"gcd({a}, {b}) = {gcd(a, b)}")
```

你可以假设输⼊将是⼀个⾮负整数，所以你不需要在你的函数中包含错误检查。例

3. 定义⼀个名为euler_approx的函数，它接受⼀个单⼀的参数，⼀个浮点数 ，并使
⽤公式（1）中的和来获得⼀个近似e的值，该值在 内接近真实值e。提⽰：使⽤
while循环。注意：你可以使⽤Python的math模块来获得e的真实值（直到浮点精
度）：math.exp(1)。
4. 定义⼀个名为print_euler_sum_table的函数，它接受⼀个单⼀的正整数 作为
参数，并打印从euler_infinite_sum(k)获得的值， 的范围从1到 ，每⾏⼀
个。


```python
import math

def euler_limit(n):
    return (1 + 1 / n) ** n

def euler_infinite_sum(n):
    total, fact = 0.0, 1
    for k in range(n):
        if k > 0:
            fact *= k
        total += 1 / fact
    return total

def euler_approx(eps):
    total, fact, k = 0.0, 1, 0
    while abs(total - math.exp(1)) >= eps:
        if k > 0:
            fact *= k
        total += 1 / fact
        k += 1
    return total

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

print("真实值 math.exp(1) =", math.exp(1))

for n in (1, 10, 100, 1000):
    print(f"euler_limit({n}) = {euler_limit(n)}")

for n in (0, 1, 4, 10):
    print(f"euler_infinite_sum({n}) = {euler_infinite_sum(n)}")

for eps in (1e-3, 1e-6, 1e-10):
    print(f"euler_approx({eps}) = {euler_approx(eps)}")

print("print_euler_sum_table(5) 输出：")
print_euler_sum_table(5)
```

测试整数属性
1. 编写⼀个名为is_power_of_2的函数，它接受⼀个正整数作为其唯⼀的参数，并返
回⼀个布尔值，指⽰输⼊是否是2的幂。你可以假设输⼊是⼀个正整数。你不能在
你的解决⽅案中使⽤内置的math.sqrt函数。你应该只需要使⽤除法和模（%）运
算。
2. 将你之前的解决⽅案推⼴到⼀个名为is_power的函数，它接受两个正整数作为其
参数， 和 ，按那个顺序，并返回⼀个布尔值。is_power(b,n)如果 是 的幂
则返回True，否则返回False。


```python
def is_power_of_2(n):
    while n % 2 == 0 and n > 1:
        n //= 2
    return n == 1

def is_power(b, n):
    if b == 1:
        return n == 1
    while n % b == 0 and n > 1:
        n //= b
    return n == 1

for n in (1, 2, 3, 4, 16, 18, 1024):
    print(f"is_power_of_2({n}) = {is_power_of_2(n)}")

for b, n in [(2, 1), (2, 8), (3, 9), (3, 10), (5, 125), (1, 1), (1, 5)]:
    print(f"is_power({b}, {n}) = {is_power(b, n)}")
```
