字符串的乐趣

在这个问题中，您将实现⼀些⽤于处理字符串的简单函数。您不需要为这个问题中的任
何函数执⾏错误检查。
1. 回⽂是指⼀个词或短语，它从前往后读和从后往前读都⼀样。例如，level、
kayak 和 pop 都是英⽂回⽂；如果忽略空格和标点符号，rats live on no
evil star 和 Was it a car or a cat I saw? 也是回⽂短语。编写⼀个名
为 is_palindrome 的函数，它接受⼀个字符串作为唯⼀参数，并返回布尔值：如
果输⼊是回⽂，返回 True，否则返回 False。您可以假设输⼊仅包含英⽂字⺟
（不区分⼤⼩写）、数字 0 到 9 和空格。判断回⽂时应忽略空格和⼤⼩写，因此
tacocat 和 T A C O cat 都应被认为是回⽂。
2. 如果⼀个词的字⺟从左到右按字⺟顺序排列，就称它是“字⺟顺序的”；允许字⺟重
复。例如，adder 和 beet 是字⺟顺序的，dog 和 cat 则不是。编写⼀个名为
is_abecedarian 的函数，它只接受⼀个字符串参数。如果参数符合上述要求，
返回 True，否则返回 False。您可以假设输⼊仅由字⺟和空格组成。判断时应忽
略空格，因此 abcd efgh xyz 被认为是字⺟顺序的。
3. 编写⼀个名为 remove_vowels 的函数，它接受⼀个字符串作为唯⼀参数，并返回
去掉所有元⾳后的字符串。本题的元⾳是字⺟ a、e、i、o、u。例如，
remove_vowels('cat') 应返回 'ct'，remove_vowels('audio') 应返回
'd'。函数也应正确处理全部由元⾳组成的字符串。

```python
def is_palindrome(s):
    # 第 1 步：去掉所有空格，并把字母统一成小写
    cleaned = s.replace(" ", "").lower()
    # 第 2 步：用切片 [::-1] 取得逆序字符串，再与原字符串比较
    # 相同则为回文，返回 True；否则返回 False
    return cleaned == cleaned[::-1]

# ---- 测试 ----
print(is_palindrome("level"))                    # True（单回文词）
print(is_palindrome("tacocat"))                  # True（单回文词）
print(is_palindrome("T A C O cat"))              # True（含空格且大小写混排）
print(is_palindrome("rats live on no evil star"))  # True（回文短语）
print(is_palindrome("kayak"))                    # True
print(is_palindrome("hello"))                    # False（不是回文）
```

```python
def is_abecedarian(s):
    # 第 1 步：去掉空格并统一转小写，避免大小写干扰比较
    cleaned = s.replace(" ", "").lower()
    # 第 2 步：把相邻的两个字符依次比较
    for i in range(len(cleaned) - 1):
        # 一旦出现"前一个字母比后一个字母大"，说明不是字母顺序
        if cleaned[i] > cleaned[i + 1]:
            return False
    # 所有相邻字符都满足 <=，则是字母顺序（允许相等/重复）
    return True

# ---- 测试 ----
print(is_abecedarian("adder"))          # True（a d d e r 递增）
print(is_abecedarian("beet"))           # True（b e e t 递增，e 重复允许）
print(is_abecedarian("dog"))            # False（d < o 但 o > g）
print(is_abecedarian("cat"))            # False（c > a）
print(is_abecedarian("abcd efgh xyz"))  # True（忽略空格）
print(is_abecedarian("a"))              # True（单字符）
print(is_abecedarian(""))               # True（空串）
```

```python
def remove_vowels(s):
    # 先把所有元音（含大写）放进一个字符串，方便用 in 判断
    vowels = "aeiouAEIOU"
    # 用空字符串逐步拼接"非元音"字符
    result = ""
    for ch in s:
        # 只有不是元音的字符才保留
        if ch not in vowels:
            result += ch
    # 如果 s 全由元音组成，result 会是空字符串 ""
    return result

# ---- 测试 ----
print(remove_vowels("cat"))          # ct
print(remove_vowels("audio"))        # d
print(remove_vowels("aeiou"))        # 空字符串（全部是元音）
print(remove_vowels("hello world"))  # hll wrld
```

列表的乐趣

在这个问题中，您将实现⼀些简单的列表操作。
1. 编写⼀个名为 list_reverse 的函数，它接受⼀个列表作为参数，并返回反转后
的列表。例如，输⼊ [1, 2, 3]，应返回 [3, 2, 1]。如果输⼊不是列表，函数
应引发适当的错误。
2. 编写⼀个名为 is_sorted 的函数，它接受⼀个序列 seq 作为唯⼀参数。如果序列
按升序排列，返回 True，否则返回 False。您可以假设 seq 是 Python 序列，且
所有元素都⽀持⽐较操作，因此⽆需检查输⼊类型。函数只应遍历序列⼀次；需要
多次遍历的低效解法不能获得满分。这个问题展⽰了 Python 动态类型的特点：只
要 seq ⽀持索引，且元素⽀持⽐较，就⽆需预先知道序列的具体类型。

```python
def list_reverse(lst):
    # 先检查类型：只有列表才允许处理
    # 传入元组、字符串等非列表时，抛出 TypeError 并说明实际类型
    if not isinstance(lst, list):
        raise TypeError(f"参数必须是列表，而不是 {type(lst).__name__}")
    # 切片 [::-1] 生成一个元素顺序反转的新列表
    # 不会修改原来的列表
    return lst[::-1]

# ---- 测试：正常反转 ----
print(list_reverse([1, 2, 3]))        # [3, 2, 1]
print(list_reverse([1, 2, 3, 4]))     # [4, 3, 2, 1]
print(list_reverse([]))               # []（空列表）
print(list_reverse(["a", "b", "c"]))  # ['c', 'b', 'a']

# ---- 测试：非法输入应抛出 TypeError ----
try:
    list_reverse((1, 2, 3))           # 元组不是列表
except TypeError as e:
    print("TypeError:", e)

try:
    list_reverse("abc")               # 字符串不是列表
except TypeError as e:
    print("TypeError:", e)
```

```python
def is_sorted(seq):
    # 只遍历一次：用下标 i 访问相邻的两个元素 seq[i] 和 seq[i+1]
    # 通过索引访问，因此列表、字符串、元组等序列都能直接用
    for i in range(len(seq) - 1):
        # 只要发现前一个元素大于后一个元素，就说明不是升序
        if seq[i] > seq[i + 1]:
            return False
    # 所有相邻元素都满足 <=（相等也算升序）
    return True

# ---- 测试 ----
print(is_sorted([1, 2, 3]))          # True
print(is_sorted([1, 3, 2]))          # False（3 > 2）
print(is_sorted([1, 2, 2, 3]))       # True（允许重复）
print(is_sorted([5]))                # True（只有一个元素）
print(is_sorted([]))                 # True（空序列）
print(is_sorted("abc"))              # True（字符串也是序列）
print(is_sorted("acb"))              # False
print(is_sorted((1, 2, 3, 0)))       # False（元组也可用）
```

作为向量的元组

在这个问题中，我们使⽤元组来表⽰向量。这⾥先⽤⼀个⻓度为d的元组表⽰d维向量。

1. 实现⼀个名为 vec_scalar_mult 的函数。它接受两个参数：⼀个由整数或浮点数组成的元组 t，以及⼀个整数或浮点数 s。函数返回与 t ⻓度相同的元组，每个元素等于 t 的对应元素乘以 s。函数应检查输⼊类型；类型不合适时，引发带有适当错误消息的 TypeError。整数标量和含有整数元素的元组都是有效输⼊。
2. 实现⼀个名为 vec_inner_product 的函数。它接受两个由整数或浮点数组成的元组，返回这两个向量内积对应的浮点数。向量 的内积是 。若任⼀输⼊不是元组，应引发 TypeError；若两个元组的⻓度不⼀致，应引发ValueError。
3. 将上述表⽰⽅法扩展到矩阵。⽤⼀个元组的元组表⽰ 矩阵 my_mx：外层元组⻓度为 ，每个内层元组⻓度为 ，表⽰矩阵的⼀⾏。编写⼀个名为check_valid_mx 的函数，它接受⼀个参数；若该参数是有效矩阵则返回 True，否则返回 False。有效矩阵满⾜：外层元组的每个元素本⾝都是元组；所有内层元组⻓度相同；每个内层元组的每个元素都是整数或浮点数。
4. 编写⼀个名为 mx_vec_mult 的函数，它接受⼀个矩阵（元组的元组）和⼀个向量（元组），返回向量与矩阵相乘得到的数值元组。函数check_valid_mx应检查参数是否合理；对于不合理的输⼊，应引发适当的错误。

```python
def vec_scalar_mult(t, s):
    # 检查第 1 个参数：必须是元组（不能是列表）
    if not isinstance(t, tuple):
        raise TypeError(f"向量必须是元组，而不是 {type(t).__name__}")
    # 检查第 2 个参数：整数或浮点数都可以作为标量
    if not isinstance(s, (int, float)):
        raise TypeError(f"标量必须是整数或浮点数，而不是 {type(s).__name__}")
    # 检查元组中的每个元素也必须是整数或浮点数
    for x in t:
        if not isinstance(x, (int, float)):
            raise TypeError(f"向量元素必须是整数或浮点数，而不是 {type(x).__name__}")
    # 生成器表达式：每个元素乘以标量 s，再用 tuple() 转回元组
    # 返回新元组，长度与 t 相同
    return tuple(x * s for x in t)

# ---- 测试：合法输入 ----
print(vec_scalar_mult((1, 2, 3), 2))       # (2, 4, 6)
print(vec_scalar_mult((1.5, 2.5), 4))      # (6.0, 10.0)
print(vec_scalar_mult((1, 2), 0.5))        # (0.5, 1.0)
print(vec_scalar_mult((), 3))              # ()（空向量）

# ---- 测试：非法输入应抛出 TypeError ----
for args in [([1, 2, 3], 2), ((1, 2), "a"), ((1, "x"), 2)]:
    try:
        vec_scalar_mult(*args)
    except TypeError as e:
        print("TypeError:", e)
```

```python
def vec_inner_product(t1, t2):
    # 检查类型：任一个不是元组都抛出 TypeError
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个参数都必须是元组")
    # 检查长度：内积要求两个向量长度相同，否则抛出 ValueError
    if len(t1) != len(t2):
        raise ValueError(f"两个向量长度不一致：{len(t1)} != {len(t2)}")
    # 检查元素：所有分量都必须是整数或浮点数
    for x in t1 + t2:
        if not isinstance(x, (int, float)):
            raise TypeError(f"向量元素必须是整数或浮点数，而不是 {type(x).__name__}")
    # zip(t1, t2) 把两个向量的对应元素两两配对
    # 逐对相乘后累加到 total 中
    total = 0.0
    for a, b in zip(t1, t2):
        total += a * b
    # 题目要求返回浮点数
    return float(total)

# ---- 测试：合法输入 ----
print(vec_inner_product((1, 2, 3), (4, 5, 6)))   # 1*4+2*5+3*6 = 32.0
print(vec_inner_product((1.5, 2.0), (2, 4)))     # 1.5*2+2*4 = 11.0
print(vec_inner_product((), ()))                 # 0.0（空向量）

# ---- 测试：非法输入 ----
try:
    vec_inner_product([1, 2, 3], (1, 2, 3))      # 列表不是元组
except TypeError as e:
    print("TypeError:", e)

try:
    vec_inner_product((1, 2), (1, 2, 3))         # 长度不一致
except ValueError as e:
    print("ValueError:", e)
```

```python
def check_valid_mx(mx):
    # 规则 1：外层必须是元组
    if not isinstance(mx, tuple):
        return False
    # 规则 1 和 3：每个元素必须是元组，且元组里的每个元素是数字
    for row in mx:
        if not isinstance(row, tuple):
            return False
        for x in row:
            if not isinstance(x, (int, float)):
                return False
    # 规则 2：所有内层元组长度必须相同
    # 以第一行的长度作为列数基准
    if len(mx) > 0:
        n_cols = len(mx[0])
        for row in mx:
            if len(row) != n_cols:
                return False
    # 三条规则都满足（空元组 () 视为 0 行矩阵）
    return True

# ---- 测试：合法矩阵 ----
print(check_valid_mx(((1, 2), (3, 4))))            # True
print(check_valid_mx(((1.0, 2.5, 3), (4, 5, 6))))  # True（含浮点数）
print(check_valid_mx(()))                          # True（0 行矩阵）

# ---- 测试：非法矩阵 ----
print(check_valid_mx([[1, 2], [3, 4]]))            # False（外层是列表）
print(check_valid_mx(((1, 2), (3, 4, 5))))         # False（行长度不一致）
print(check_valid_mx(((1, 2), (3, "x"))))          # False（元素不是数字）
print(check_valid_mx(((1, 2), 3)))                 # False（内层不是元组）
print(check_valid_mx(5))                           # False（根本不是元组）
```

```python
def mx_vec_mult(mx, v):
    # 第一步：用 check_valid_mx 校验矩阵是否合理
    # 不合法说明参数类型错误，抛出 TypeError
    if not check_valid_mx(mx):
        raise TypeError("第一个参数不是有效的矩阵")
    # 第二步：校验向量本身必须是元组
    if not isinstance(v, tuple):
        raise TypeError(f"向量必须是元组，而不是 {type(v).__name__}")
    # 第三步：向量中的每个元素也必须是数字
    for x in v:
        if not isinstance(x, (int, float)):
            raise TypeError(f"向量元素必须是整数或浮点数，而不是 {type(x).__name__}")
    # 空矩阵（0 行）没有行可以做内积，直接返回空元组
    if len(mx) == 0:
        return ()
    # 第四步：矩阵列数（每行长度）必须等于向量长度，否则无法相乘
    n_cols = len(mx[0])
    if len(v) != n_cols:
        raise ValueError(f"向量长度 {len(v)} 与矩阵列数 {n_cols} 不一致")
    # 第五步：把矩阵的每一行看作一个向量，与 v 做内积
    # 结果元组的第 i 个分量 = 第 i 行与向量 v 的内积
    return tuple(vec_inner_product(row, v) for row in mx)

# ---- 测试：合法输入 ----
my_mx = (
    (1, 2),
    (3, 4),
)
print(mx_vec_mult(my_mx, (5, 6)))        # (17.0, 39.0)
print(mx_vec_mult(my_mx, (1.0, 0.5)))    # (2.0, 5.0)
print(mx_vec_mult((), ()))               # ()

# ---- 测试：非法输入 ----
try:
    mx_vec_mult([[1, 2], [3, 4]], (5, 6))   # 矩阵不是元组的元组
except TypeError as e:
    print("TypeError:", e)

try:
    mx_vec_mult(my_mx, [5, 6])              # 向量不是元组
except TypeError as e:
    print("TypeError:", e)

try:
    mx_vec_mult(my_mx, (5, 6, 7))           # 维度不匹配
except ValueError as e:
    print("ValueError:", e)
```

更多向量的乐趣

在这个练习中，我们将再次遇到向量，这次采⽤⾯向对象的⽅法。
1. 定义⼀个名为 Vector 的类。每个向量都有⼀个维度（⾮负整数）和⼀个元素列表或元组。初始化函数的第⼀个参数是维度，第⼆个参数是由整数或浮点数组成的列表或元组，表⽰向量的元素。如果只提供维度⽽不提供元素，请选择合理的默认⾏为。若维度的类型不正确或为负数，应抛出合适的异常；若维度与元素个数不⼀致，也应抛出异常。
2. 维度和元素应是类属性还是实例属性？说明选择的理由。
3. 实现向量对象的⽐较运算符，包括相等、⼩于、⼩于或等于、⼤于等。两个Vector 对象的坐标相同，则视为相等；否则按 Python 元组的字典序⽐较，即先⽐较第⼀个坐标，再⽐较第⼆个坐标，依此类推。例如，⼆维向量 (2, 4) ⼩于(2, 5)。⽐较维度不同的两个向量时，应抛出异常。
4. 实现 Vector.dot ⽅法，接收另⼀个 Vector 对象，返回两向量的内积。若参数
类型不正确或维度不⼀致，应抛出合适的异常。
5. 实现 Vector 的乘法运算符。标量左乘或右乘（如 2*v 和 v*2）应返回⼀个各元素都乘以该标量的新 Vector 对象。当两个 Vector 对象相乘（v*w）时，对应位置的元素逐⼀相乘，结果仍是⼀个新 Vector 对象；例如，Vector(2, (2, 3))* Vector(2, (4, 5)) 的元素为 (8, 15)。若两个向量的维度不⼀致，应抛出合适的异常。

```python

class Vector:
    """d 维向量：维度 dim + 元素列表 elems。"""

    def __init__(self, dim, elems=None):
        # 维度必须是整数
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError(f"维度必须是整数，而不是 {type(dim).__name__}")
        # 维度不能为负
        if dim < 0:
            raise ValueError(f"维度必须是非负整数，而不是 {dim}")

        # 只给维度时，默认创建零向量
        if elems is None:
            elems = [0] * dim

        # 元素必须是列表或元组
        if not isinstance(elems, (list, tuple)):
            raise TypeError(f"元素必须是列表或元组，而不是 {type(elems).__name__}")

        # 每个元素必须是整数或浮点数
        for x in elems:
            if not isinstance(x, (int, float)) or isinstance(x, bool):
                raise TypeError(f"元素必须是整数或浮点数，而不是 {type(x).__name__}")

        # 维度必须与元素个数一致
        if len(elems) != dim:
            raise ValueError(f"维度 {dim} 与元素个数 {len(elems)} 不一致")

        # 维度和元素都保存为实例属性（复制列表，避免外部修改影响对象）
        self.dim = dim
        self.elems = list(elems)

    def __repr__(self):
        return f"Vector({self.dim}, {tuple(self.elems)})"


# ---- 测试 ----
print(Vector(3, (1, 2, 3)))
print(Vector(3, [1.5, 2.5, 3.5]))
print(Vector(2))  # 默认零向量

for args in [(-1, ()), ("3", (1, 2, 3)), (2, (1, 2, 3)), (2, "ab"), (2, (1, "x"))]:
    try:
        Vector(*args)
    except (TypeError, ValueError) as e:
        print(type(e).__name__, ":", e)
```

维度和元素应是类属性还是实例属性？说明选择的理由。

都应当是实例属性。

理由：
1. 每个向量的维度和元素都不同，属于"每个对象各自的数据"。
用 self.dim / self.elems 保存，各实例互不影响。
2. 若定义为类属性，则所有实例共享同一个值（元素还是同一个列表），
修改一个向量的元素会"连坐"地改掉其他向量的元素，违背对象独立性。
3. 类属性适合存放所有实例共享的常量或行为（如类级常量、@staticmethod），
而"这个向量自身有几个分量、分量是多少"显然属于单个对象的状态。
